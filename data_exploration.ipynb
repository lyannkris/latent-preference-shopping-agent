```python
import json

file_name = "data/items_ins_v2_filtered.json"

with open(file_name, "r") as f:
    data = json.load(f)

attributes = {}

for i in data:
    for att in data[i]["attributes"]:
        if att in attributes:
            attributes[att] += 1
        else:
            attributes[att] = 1
```
