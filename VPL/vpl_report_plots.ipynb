# VPL Report Plots

This notebook loads the current VPL result files and generates report figures for the 10,000-product setting.

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt

BASE = Path('/Users/ee/Documents/UMich/Course/2026WN/EECS545/Project/20260417/20260422/VPL')
RESULTS = BASE / 'results_embeddings10000'
OUT = BASE

with open(RESULTS / 'train_history.json', 'r', encoding='utf-8') as f:
    train_history = json.load(f)

with open(RESULTS / 'active_test_results.json', 'r', encoding='utf-8') as f:
    active_results = json.load(f)

with open(RESULTS / 'test_results.json', 'r', encoding='utf-8') as f:
    fixed_results = json.load(f)

train_history.keys(), active_results.keys(), fixed_results.keys()
```

```python
# Plot 1: training and validation regret by epoch
epochs = np.arange(1, len(train_history['eval_avg_regret']) + 1)
plt.figure(figsize=(7, 4.5))
plt.plot(epochs, train_history['train_loss'], label='Train Loss', linewidth=2)
plt.plot(epochs, train_history['eval_avg_regret'], label='Validation Regret', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Value')
plt.title('VPL Training and Validation Regret by Epoch')
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(OUT / 'vpl_training_regret.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# Plot 2: posterior entropy by turn
entropy_by_turn = np.array([ep['posterior_entropy_by_turn'] for ep in active_results['per_episode']])
mean_entropy = entropy_by_turn.mean(axis=0)
turns = np.arange(1, len(mean_entropy) + 1)

plt.figure(figsize=(7, 4.5))
plt.plot(turns, mean_entropy, marker='o', linewidth=2)
plt.xlabel('Interaction Turn')
plt.ylabel('Average Posterior Entropy')
plt.title('VPL Posterior Uncertainty Reduction Across Pairwise Interactions')
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(OUT / 'vpl_uncertainty_reduction.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# Plot 3: regret distribution
regrets = np.array([ep['regret'] for ep in active_results['per_episode']])
plt.figure(figsize=(7, 4.5))
plt.hist(regrets, bins=20, edgecolor='black', alpha=0.8)
plt.axvline(regrets.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean = {regrets.mean():.4f}')
plt.xlabel('Regret')
plt.ylabel('Count')
plt.title('VPL Regret Distribution on the Test Set')
plt.legend()
plt.grid(alpha=0.15)
plt.tight_layout()
plt.savefig(OUT / 'vpl_regret_distribution.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# Plot 4: success at k
metrics = active_results['metrics']
k_values = [1, 3, 5, 10, 50, 100, 200, 500, 1000]
success_values = [
    metrics['success_at_1'], metrics['success_at_3'], metrics['success_at_5'],
    metrics['success_at_10'], metrics['success_at_50'], metrics['success_at_100'],
    metrics['success_at_200'], metrics['success_at_500'], metrics['success_at_1000']
]

plt.figure(figsize=(7, 4.5))
plt.plot(k_values, success_values, marker='o', linewidth=2)
plt.xscale('log')
plt.xticks(k_values, [str(k) for k in k_values])
plt.xlabel('K')
plt.ylabel('Success@K')
plt.title('VPL Success at K')
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(OUT / 'vpl_success_at_k.png', dpi=200, bbox_inches='tight')
plt.show()
```
